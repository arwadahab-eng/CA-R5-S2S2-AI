```python
grade_degree = float(input("enter your degree"))
if 0 <= grade_degree <= 100:
    if 90 <= grade_degree <= 100:
        grade_degree = "A"
        
    elif 80 <= grade_degree <= 90:
        grade_degree = "B"
        
    elif 70 <= grade_degree <= 80:
        grade_degree = "C"
        
    elif 70 <= grade_degree <= 60:
        grade_degree = "D" 
           
    elif grade_degree < 60:
        grade_degree = "F" 

       
else :
    print("try enter your grade again")
    

print(grade_degree)
```
