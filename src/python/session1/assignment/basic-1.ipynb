```python
import re

text = "###!!@emocleW EPGTQ!!!6789"

# Keep only letters and spaces
cleaned = re.sub(r'[^a-zA-Z\s]', '', text)   # 'emocleW EPGTQ'

words = cleaned.split()   # ['emocleW', 'EPGTQ']

# Remove any 'E'/'e' characters, then reverse each word
result = " ".join(re.sub(r'[E]', '', w)[::-1] for w in words)

print(result)
```
